# Data Binning with Pandas
Binning allows us to group continuous data into intervals. This notebook covers:
1. **Fixed-width binning** (Defining your own edges)
2. **Equal-width binning** (Automatic linear splitting)
3. **Quantile binning** (Equal number of observations per bin)

```python
import pandas as pd
import numpy as np

# Sample data: Exam scores for 12 students
df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J', 'K', 'L'],
    'Score': [45, 55, 62, 72, 78, 81, 85, 88, 91, 95, 98, 100]
})
df
```

### 1. Fixed-Width Binning (pd.cut)
Assigning custom letter grades based on specific score ranges.

```python
bins = [0, 60, 75, 90, 100]
labels = ['Fail', 'C', 'B', 'A']

df['Grade'] = pd.cut(df['Score'], bins=bins, labels=labels)
df
```

### 2. Equal-Width Binning
Pandas calculates the edges automatically to create 3 bins of equal range length.

```python
df['Range_Group'] = pd.cut(df['Score'], bins=3)
df.groupby('Range_Group').size()
```

### 3. Equal-Frequency Binning (pd.qcut)
Splitting data so each bin has the same number of data points (Quantiles).

```python
df['Quantile_Group'] = pd.qcut(df['Score'], q=4, labels=['Q1', 'Q2', 'Q3', 'Q4'])
df
```
