## Aggregation

```python
# dataframe with nba players, their teams, and their points per game, and country of origin
import pandas as pd
data = {
    'Player': ['LeBron James', 'Kevin Durant', 'Stephen Curry', 'Giannis Antetokounmpo', 'Kawhi Leonard'],
    'Team': ['Los Angeles Lakers', 'Brooklyn Nets', 'Golden State Warriors', 'Milwaukee Bucks', 'Los Angeles Clippers'],
    'Points Per Game': [25.3, 27.1, 30.2, 28.5, 24.8],
    'Country of Origin': ['USA', 'USA', 'USA', 'Greece', 'USA']
}
df = pd.DataFrame(data)
```

## get average

```python
# get average points per game for all players
average_points_all = df['Points Per Game'].mean()
average_points_all
```

```python
# filter for USA players
usa_players = df[df['Country of Origin'] == 'USA']
print(usa_players)
```

```python
# get average points per game for USA players
average_points_usa = usa_players['Points Per Game'].mean()
average_points_usa
```

## using groupby

```python
# use groupby to get average points per game by country of origin
average_points_by_country = df.groupby('Country of Origin')['Points Per Game'].mean()
average_points_by_country
```
