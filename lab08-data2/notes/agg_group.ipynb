# Pandas Grouping and Aggregation Demo
This notebook demonstrates how to use the `.groupby()` and `.agg()` functions to summarize data efficiently.

```python
import pandas as pd

# Creating a sample dataset
data = {
    'Department': ['Sales', 'Sales', 'IT', 'IT', 'HR', 'Sales', 'IT', 'HR'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Edward', 'Fiona', 'George', 'Hannah'],
    'Salary': [70000, 80000, 90000, 85000, 60000, 75000, 95000, 65000],
    'Experience_Years': [5, 7, 3, 4, 2, 6, 8, 3]
}

df = pd.DataFrame(data)
df
```

```python
df['Salary'].mean()
```

### 1. Basic Grouping
Calculate the mean salary for each department.

```python
df.groupby('Department')['Salary']
```

```python
avg_salary = df.groupby('Department')['Salary'].mean()
print(avg_salary)
```

### 2. Multiple Aggregations
You can perform multiple calculations at once using `.agg()`.

```python
summary = df.groupby('Department').agg({
    'Salary': ['mean', 'max'],
    'Experience_Years': 'sum'
})

# hierarchical index
# summary['Salary']['mean'].loc['HR']
summary
```

### stacking and unstacking

```python
ss = summary.stack().stack()
ss
# ss.index

# # ss.loc['HR'].loc['max']
# ss.loc[('HR','max')]
```

### 3. Named Aggregation
This is a cleaner way to rename your columns while aggregating.

```python
named_agg = df.groupby('Department').agg(
    total_payroll=('Salary', 'sum'),
    average_experience=('Experience_Years', 'mean')
)
named_agg
```
