# Python Fundamentals: Lists and Loops
## Processing Collections of Data
This notebook covers how to store multiple values and iterate through them using loops.

### 1. Python Lists
A **list** is an ordered collection of items. In data science, we use lists to store features, observations, or results.

```python
# Creating a list of transaction sizes
transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

# Accessing items (Indexing starts at 0)
print(f"First transaction: {transactions[0]}")
print(f"Number of transactions: {len(transactions)}")
```

### 2. For Loops
A `for` loop allows you to execute a block of code once for each item in a list.

```python
print("Processing transactions:")
for amount in transactions:
    if amount > 100:
        print(f"High value: ${amount}")
    else:
        print(f"Standard value: ${amount}")
```

### Building a list

```python
transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

fee_rate = 0.10
with_fee_applied = []

for t in transactions:
    with_fee_applied.append(t*fee_rate)

with_fee_applied
```

```python
# using a generator

transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

fee_rate = 0.10
with_fee_applied = [t*fee_rate for t in transactions]

with_fee_applied
```

### 3. Combining Functions with Loops
We can use the logic from previous labs to transform entire lists. This is a common pattern in **Data Normalization**.

```python
def min_max_scaler(value, min_val, max_val):
    if max_val == min_val:
        return 0.0
    return (value - min_val) / (max_val - min_val)

raw_data = [10, 20, 30, 40, 50]
scaled_data = []

min_v = min(raw_data)
max_v = max(raw_data)

for x in raw_data:
    scaled_x = min_max_scaler(x, min_v, max_v)
    scaled_data.append(scaled_x)

print(f"Original: {raw_data}")
print(f"Scaled:   {scaled_data}")
```
