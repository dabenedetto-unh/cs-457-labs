### Type checking

```python
x = 5

print(isinstance(x, int))
print(isinstance(x, float))
print(isinstance(x, (int,float)))
```

### Looping

```python
data = [1, 2, 3.5, 4]

all_numeric = True
for x in data:
    if not isinstance(x,(int,float)):
        all_numeric = False
        break   # stop looping, we're done

print(all_numeric)
```

### Generators

```python
# basic generator

data = [1, 2, 3, 4]

doubles = [2*x for x in data]
doubles
```

* Now we can use our type check with the generator

```python
data = [1, 2, "three", 4]

bools = [isinstance(x, (int, float)) for x in data]
bools
```

* Finally, we can use 'all', which will return True only if all of the values in the list are True

```python
data = [1, 2, "three", 4]

# Check for both integers and floats
is_numeric = all(isinstance(x, (int, float)) for x in data)

print(is_numeric) # Returns: True
```
