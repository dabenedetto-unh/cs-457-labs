# Python Fundamentals: Text, Mappings, and Persistence
## Handling Strings, Dictionaries, and Files
This notebook covers how to manipulate text, store data in key-value pairs, and read/write data to external files. These concepts build upon variables and control flow.

### 1. Python Strings
A **string** is an ordered sequence of characters. Like lists, they support indexing and slicing to access specific parts of the data.

```python
# String manipulation
sample_text = "Data Science"

# Slicing: [start:stop] (Indexing starts at 0)
print(f"First word: {sample_text[0:4]}") 

# Common String Methods
print(sample_text.upper())    # Convert to uppercase
print(sample_text.lower())    # Convert to lowercase
print("1,2,3".split(","))     # Split into a list
```

### 2. Dictionaries (`dict`)
A **dictionary** is a collection that stores data in key-value pairs. Unlike lists, which use numeric indices starting at 0, dictionaries use unique keys to retrieve values.

```python
# Creating a dictionary for a user profile
user_profile = {
    "user_id": 101,
    "username": "jdoe_data",
    "access_level": "admin"
}

# Accessing values via keys
print(f"Username: {user_profile['username']}")

# Adding or updating items
user_profile["last_login"] = "2026-01-24"

# Iterating through a dictionary using a loop
for key, value in user_profile.items():
    print(f"{key}: {value}")
```

### 3. File Input/Output (I/O)
File I/O allows your program to persist data by reading from or writing to files. The `with` statement is the standard way to handle files safely, ensuring they are closed after the block executes.

```python
# 1. Writing to a file
log_entries = ["System Start", "Analysis Complete", "Report Generated"]

with open("activity_log.txt", "w") as file:
    for entry in log_entries:
        file.write(f"{entry}\n")
print("File 'activity_log.txt' has been created.")

# 2. Reading from a file
print("\nReading log contents:")
with open("activity_log.txt", "r") as file:
    for line in file:
        print(f"Log Entry: {line.strip()}")
```

### 4. Practical Application: Data Normalization and Parsing
Combining loops and strings to process raw data formats. This follows the pattern of using functions to transform data collections.

```python
raw_data = ["id,score", "1,85", "2,92", "3,78"]
processed_data = {}

for line in raw_data[1:]:  # Slicing to skip the header
    parts = line.split(",")
    s_id = int(parts[0])
    score = int(parts[1])
    processed_data[s_id] = score

print(f"Processed Scores Dictionary: {processed_data}")
```
