## 1. Python Strings
A **string** is an ordered sequence of characters. Like lists, they support indexing and slicing to access specific parts of the data.

```python
# String
sample_text = "Data Science"
sample_text
```

```python
# string concatenation
sample_text = "Data" + " " + "Science"
sample_text
```

### String access

```python
# index like a list
sample_text[1]
```

```python
# Slicing: [start:stop] (Indexing starts at 0)
print(f"First word: <{sample_text[0:4]}>") 
```

```python
# characters
for char in sample_text[0:4]:
    print(char)
```

### Transformations

```python
# Common String Methods
print(sample_text.upper())    # Convert to uppercase
print(sample_text.lower())    # Convert to lowercase
```

```python
sample_text = "D a t a"
sample_text
```

```python
# replace

sample_text = sample_text.replace(' ','')
sample_text
```

### String parsing

```python
sample_text.isdigit()

print("12.3".isdigit())
print()
for c in "12.3":
    print(c.isdigit())
```

```python
# parsing strings

s = "123"
i = int(s)

print(s + s)
print(i + i)
```

```python
try:
    int("1 2 3")
except ValueError as e:
    print(e)
```

### Split and join

```python
"1,2,3".split(",")     # Split into a list
```

```python
",".join(['1', '2', '3'])   # inverse of split
```
