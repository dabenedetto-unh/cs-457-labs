```python
import numpy as np

# Generate a normal distribution dataset
# np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=0, scale=1, size=1000)

print(f"Generated {len(data)} samples from normal distribution")
print(f"Mean: {np.mean(data):.4f}, Std: {np.std(data):.4f}")
```

```python
from scipy.stats import norm

# We want the 75th percentile (0.75) to find the boundary for the middle 50%
z_score = norm.ppf(0.75)

print(z_score) 
# Output: 0.6744897501960817
```

```python
# Calculate quartiles for the data
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)  # This is the median
q3 = np.percentile(data, 75)

print(f"Q1 (25th percentile): {q1:.4f}")
print(f"Q2 (50th percentile/Median): {q2:.4f}")
print(f"Q3 (75th percentile): {q3:.4f}")
print(f"IQR (Interquartile Range): {q3 - q1:.4f}")
```
