```python
import pandas as pd
import plotly.express as px
```

```python
data = {'state_code': ['NY', 'CA', 'TX', 'FL', 'WA'],
        'score': [100, 200, 300, 400, 500]}
df_states = pd.DataFrame(data)
df_states.head()
```

```python
# # 1. Use a built-in dataset (or any DF with 2-letter state codes)
# df = px.data.gapminder().query("country == 'United States'") 

# 2. Plot without a GeoJSON
fig = px.choropleth(
    df_states,
    locations='state_code',      # Column with state abbreviations
    locationmode="USA-states",   # Tells Plotly to use built-in US geometries
    scope="usa",                 # Zooms the map to the USA
    color='score',
    title='US State Built-in Example',
    color_continuous_scale="Viridis"
)

fig.show()
```
