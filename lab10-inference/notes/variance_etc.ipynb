## Variance, stdev, and stderr

For populations, sample means, and differences between sample means

n is sample size

**For a population:**

$$\text{variance}=\sigma^2$$
$$\text{stdev}=\sigma$$

**For sample means:**

$$\text{variance}=\frac{\sigma^2}{n}$$
$$\text{stderr}=\frac{\sigma}{\sqrt{n}}$$

**For difference between sample means**

$$\text{variance}=\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}$$
$$\text{stderr}=\sqrt{\text{variance}}
