```python
import pandas as pd
# import wbdata
```

```python
df = pd.read_csv('data/.notshared/essential_indicators.csv')
df.head()
```

```python
df = df.fillna("-")
df.head()
```

```python
if type(df.iloc[0]['GDP_USD']) is float:

    # add $ signs
    mask = df['GDP_USD'] != "-"
    df.loc[mask, 'GDP_USD'] = '$' + df['GDP_USD'].loc[mask].astype(str)

    mask = df['GDP_per_Capita'] != "-"
    df.loc[mask, 'GDP_per_Capita'] = '$' + df['GDP_per_Capita'].astype(str)

    mask = df['Poverty_Rate'] != "-"
    df.loc[mask, 'Poverty_Rate'] = (df['Poverty_Rate']*100).astype(str) + '%'

df.head()
```

```python
# # replace periods with commas
# Replaces all '.' with ',' and returns a new DataFrame of strings
df_formatted = df.astype(str).replace(r'\.', ',', regex=True)
df_formatted.to_csv('data/essential_indicators_messy.csv')
```
