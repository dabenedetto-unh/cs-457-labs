```python
import numpy as np
import pandas as pd
import seaborn as sns
import faker

fake = faker.Faker()
```

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 40, 35, 30],
    'City': ['New York', 'London', 'New York', 'Tokyo']
}

df = pd.DataFrame(data)
df.head()
```

## Numeric

```python
n=100
mu = 70     # mean
sigma = 2.75    # stdev

def get_weight(h):
    w_mu = 5.7*h - 206
    w_sigma = 28

    # get random weight based on the given mu and sigma
    w = np.random.normal(loc=w_mu, scale=w_sigma)
    return w

names = [fake.first_name_male() for _ in range(n)]
heights = np.random.normal(loc=mu, scale=sigma, size=n)
weights = [get_weight(h) for h in heights]


fake_data = pd.DataFrame({
    'Name':names,
    "Height":heights,
    "Weight":weights
})

fake_data.head()
```

```python
sns.histplot(fake_data,x="Height")
```

```python
sns.scatterplot(fake_data,x="Height",y="Weight")
```

```python
cols = ['Height',"Weight"]
corr = fake_data[cols].corr()

sns.heatmap(corr, annot=True)
```

# Correlation Interpretation Guide

| Correlation Range | Strength |
|---|---|
| ±0.81 to ±0.99 | Very strong correlation |
| ±0.61 to ±0.80 | Strong correlation |
| ±0.31 to ±0.60 | Moderate correlation |
| ±0.11 to ±0.30 | Weak correlation |
| 0.00 to ±0.10 | Negligible or no correlation |
