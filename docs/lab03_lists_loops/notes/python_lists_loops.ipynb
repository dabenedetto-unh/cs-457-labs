## Tuples

A tuple is an ordered collection of items.
A tuple is immutable - once it is created, it cannot change.

```python
# example of tuple
my_tuple = (1, 2, 3, "hello", 4.5)
print("Original tuple:", my_tuple)
# accessing elements
print("First element:", my_tuple[0])
```

### 1. Python Lists
A **list** is an ordered collection of items. In data science, we use lists to store features, observations, or results.

```python
# Creating a list of transaction sizes
transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

t0 = transactions[0]
print("First transaction:",t0)
```

You can modify values in a list, or add elements to a list

```python
transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

# Modifying a list
transactions[2] = 550.75
print("Modified transactions:", transactions)

# adding or removing elements
transactions.append(250.00)
print("After appending:", transactions)

# removing
transactions.remove(120.00)

# removing element at index
transactions.pop(0)
print("After removing:", transactions)
```

```python
# inclusion testing
items = ['A','B','C','D']

print('A' in items)
print('B' in items)
print('Y' in items)

x = "A"
if x in items:
    print(x,"is in items")
```

### 2. For Loops
A `for` loop allows you to execute a block of code once for each item in a list.

```python
# print each element
for amt in transactions:
    print(amt)
```

```python
print("Processing transactions:")
for amount in transactions:
    if amount > 100:
        # print("high")
        print(f"High value: ${amount}")
    else:
        print(f"Standard value: ${amount}")

    # print("next")
```

### Building a list

```python
transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

fee_rate = 0.10
with_fee_applied = [] # start with an empty list  

for t in transactions:
    with_fee = t*(1+fee_rate)
    with_fee_applied.append(with_fee) # add to list

with_fee_applied
```

```python
# using a generator

transactions = [10.50, 120.00, 500.25, 1500.00, 5.99]

fee_rate = 0.10
with_fee_applied = [t*(1+fee_rate) for t in transactions ]

with_fee_applied
```
