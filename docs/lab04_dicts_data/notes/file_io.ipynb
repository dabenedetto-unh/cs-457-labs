### 3. File Input/Output (I/O)
File I/O allows your program to persist data by reading from or writing to files. The `with` statement is the standard way to handle files safely, ensuring they are closed after the block executes.

```python
# 1. Writing to a file
log_entries = ["System Start", "Analysis Complete", "Report Generated"]

with open("activity_log.txt", "w") as file:
    for entry in log_entries:
        # file.write(f"{entry}\n")
        file.write(entry+'\n')
print("File 'activity_log.txt' has been created.")

# 2. Reading from a file
print("\nReading log contents:")
with open("activity_log.txt", "r") as file:
    for line in file:
        # print(f"Log Entry: {line.strip()}")
        print(line.strip())
```
