## 1. Python Strings
A **string** is an ordered sequence of characters. Like lists, they support indexing and slicing to access specific parts of the data.

```python
# String
sample_text = "Data Science"
sample_text
```

```python
# string concatenation
sample_text = "Data" + " " + "Science"
sample_text
```

### String access

```python
# index like a list
sample_text[1]
```

```python
# Slicing: [start:stop] (Indexing starts at 0)
print(f"First word: {sample_text[0:4]}") 
```

```python
# characters
for char in sample_text[0:4]:
    print(char)
```

### String methods

```python
sample_text.isdigit()

print("123".isdigit())

for c in "123":
    print(c.isdigit())
```

```python
# parsing strings

s = "123"
i = int(s)

print(s + s)
print(i + i)
```

```python
try:
    int("1 2 3")
except ValueError as e:
    print(e)
```

```python
# Common String Methods
print(sample_text.upper())    # Convert to uppercase
print(sample_text.lower())    # Convert to lowercase
```

```python
# replace

sample_text.replace(' ','')
```

### Split and join

```python
"1,2,3".split(",")     # Split into a list
```
