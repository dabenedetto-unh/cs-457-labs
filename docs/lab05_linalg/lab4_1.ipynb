## Matrix multiplication

Note: Dot product is a special case of matrix multiplication, where both matrices are vectors (1D matrix)

Imagine you are a data scientist for a real estate app. You have a simple model where:

* Square Footage has a weight (importance) of 0.1 (every sq ft adds $0.1k).
* Number of Bedrooms has a weight of 50 (every bedroom adds $50k).

```python
# import numpy as np

# 1. FEATURE MATRIX (X)
# Columns = [Square Footage, Number of Bedrooms]
X = [
    [2000, 3],  # House 1
    [1500, 2],  # House 2
    [1200, 2],  # House 3
    [2500, 4]   # House 4
]

# 2. WEIGHTS VECTOR (w)
# These represent the 'impact' of each feature on the price
# Index 0: 0.1 ($100 per sq ft)
# Index 1: 50.0 ($50,000 per bedroom)
weights = [[0.1], [50.0]]

# 3. MATRIX MULTIPLICATION (X @ w)
# This calculates: (SqFt * 0.1) + (Bedrooms * 50) for every row simultaneously
# predictions = X @ weights
predictions = lab4.matrix_mult(X,weights)

# Display the results
print("House Data (Sq Ft | Bedrooms):")
print(X)
print("\nPredicted Prices (in thousands $):")
for i, price in enumerate(predictions):
    print(f"House {i+1}: ${price}k")
```
