# NumPy & Linear Algebra Fundamentals
---
This notebook covers the essential "Day 1" skills for handling vectors and matrices using Python's `numpy` library.

```python
import numpy as np
```

## 1. Vectors: Magnitude and Dot Products
Vectors are the fundamental directional data structures in linear algebra.

```python
# Create a vector
v = np.array([3, 4])

# 1. Magnitude (Norm)
mag = np.linalg.norm(v)

# 2. Dot Product
a = np.array([1, 2])
b = np.array([3, 4])
dot_val = np.dot(a, b)

print(f"Vector v: {v}")
print(f"Magnitude of v: {mag}")
print(f"Dot product of a and b: {dot_val}")
```

## 2. Matrices: Operations and Transformations
Matrices allow us to perform linear transformations on vectors.

```python
# Create a 2x2 Matrix
A = np.array([[1, 2], [3, 4]])

# Transpose (swap rows and columns)
A_transpose = A.T

# Determinant (if 0, matrix is singular/not invertible)
det_A = np.linalg.det(A)

print(f"Matrix A:\n{A}")
print(f"Transpose:\n{A_transpose}")
print(f"Determinant: {det_A:.2f}")
```

## 3. Matrix Multiplication
Remember to use `@` or `np.matmul()`. Using `*` will perform element-wise multiplication, which is usually NOT what you want in linear algebra.

```python
B = np.array([[5, 6], [7, 8]])

# Matrix-Matrix multiplication
C = A @ B

print(f"Result of A @ B:\n{C}")
```

## 4. Eigenvalues and Eigenvectors
These represent the 'characteristic' directions of a matrix where the transformation only scales the vector rather than rotating it.

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors:\n{eigenvectors}")
```
