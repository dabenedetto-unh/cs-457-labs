## Vector sum

If a plane is flying North at 500 mph (Vector A) and there is a crosswind pushing it East at 50 mph (Vector B), the plane's actual velocity is the vector sum of the two.

```python
plane_vel = np.array([0, 500])
wind = np.array([50, 0])

v_s = plane_vel + wind
v_s
```

```python
# magnitude

v_s_mag = np.linalg.norm(v_s)
v_s_mag
```

The features might represent things like ["Action/Adventure level," "Family-friendliness," "Sci-Fi focus,"] etc. (though the machine learning model doesn't label them, they are just numbers).

A user has a preference vector that represents how much they like those aspects.  Each movie has a vector that represents how much of those aspects are present in the movie.

```python
# labels
labels = ["Action/Adventure level," "Family-friendliness," "Sci-Fi focus,"]

# user preferences
U_A = np.array([ 0.9, -0.1, 0.4])

# movie attributes
M_X = np.array([0.8,0.2,0.3])
M_Y = np.array([0.1, -0.9, 0.7])

# match strength
P_X = U_A @ M_X
P_Y = U_A @ M_Y

P_X, P_Y
```

## Matrix multiplication

Note: Dot product is a special case of matrix multiplication, where both matrices are vectors (1D matrix)

Imagine you are a data scientist for a real estate app. You have a simple model where:

* Square Footage has a weight (importance) of 0.1 (every sq ft adds $0.1k).
* Number of Bedrooms has a weight of 50 (every bedroom adds $50k).

```python
# import numpy as np

# 1. FEATURE MATRIX (X)
# Columns = [Square Footage, Number of Bedrooms]
X = np.array([
    [2000, 3],  # House 1
    [1500, 2],  # House 2
    [1200, 2],  # House 3
    [2500, 4]   # House 4
])

# 2. WEIGHTS VECTOR (w)
# These represent the 'impact' of each feature on the price
# Index 0: 0.1 ($100 per sq ft)
# Index 1: 50.0 ($50,000 per bedroom)
weights = np.array([0.1, 50.0])

# 3. MATRIX MULTIPLICATION (X @ w)
# This calculates: (SqFt * 0.1) + (Bedrooms * 50) for every row simultaneously
predictions = X @ weights

# Display the results
print("House Data (Sq Ft | Bedrooms):")
print(X)
print("\nPredicted Prices (in thousands $):")
for i, price in enumerate(predictions):
    print(f"House {i+1}: ${price}k")
```
