# Pandas Essentials: A Quick-Start Guide

Pandas is the bread and butter of Python data analysis. This notebook covers the core operations you'll use 90% of the time.

```python
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
```

## Creating Data

You can create a **DataFrame** (a table) from dictionaries or lists.

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 40, 35, 30],
    'City': ['New York', 'London', 'Paris', 'Tokyo']
}

df = pd.DataFrame(data)
df.head()
```

## reading and writing CSV

```python
# write df to csv - this will write the 
df.to_csv("fake_data.csv")

# read fake_data to df_in
df_in = pd.read_csv("fake_data.csv", index_col=0)
df_in.head()
```

## Basic Exploration
Get a feel for your data's shape and statistics.

```python
print("--- Info ---")
df.info()

print("\n--- Descriptive Stats ---")
display(df.describe())
```

## Column access

Get a single column

```python
# Select a column
ages = df['Age']
ages
```

```python
# get multiple columns
cols = ['Name','City']
name_city = df[cols]
name_city
```

## Sorting & Filtering

```python
# ascending=True is default
sorted_df = df.sort_values('Age', ascending=True)
sorted_df
```

```python
# Filter rows where Age > 30
f = (df['Age'] > 30)
over_30 = df[f]
over_30
```

## Adding & Modifying Columns
Vectorized operations make this incredibly fast.

```python
# Create a new column based on logic
f = (df['Age'] >= 35)
df['Senior'] = df[f]

# Uppercase the cities
df['City'] = df['City'].str.upper()
df
```
