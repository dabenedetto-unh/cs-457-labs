### 1. Variables, Expressions, and Statements (Ch. 2)
A **variable** is a name that refers to a value. An **expression** is a combination of values, variables, and operators.

```python
# Assignment and Types
message = "Hello, Python!"  # String
n = 17                     # Integer
pi = 3.14159               # Float

new_val = n*pi
print(new_val)

print(type(message))
print(type(n))
print(type(pi))
```

### 2. Functions (Ch. 4)
Functions allow us to group a set of statements together to run them as a single unit.

```python
def calculate_area(radius):
    """Calculates the area of a circle."""
    area = 3.14159 * (radius ** 2)
    return area

result = calculate_area(5)
print(f"The area is: {result}")
```

### 3. Conditionals (Ch. 5)
Conditionals allow us to check conditions and change the behavior of the program accordingly.

```python
x = 5

if x > 0:
    print("x is positive")
elif x < 0:
    print("x is negative")
else:
    print("x is zero")
```
